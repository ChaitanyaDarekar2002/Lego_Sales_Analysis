```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('lego_sets.csv')
data
```

```python
data.info()
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
data["subtheme"] = data["subtheme"].fillna("LEGO brand stores")
```

```python
data.isnull().sum()
```

```python
data.head()
```

```python
data["pieces"] = data["pieces"].fillna("233.0")
```

```python
data.isnull().sum()
```

```python
data
```

```python
data["minifigs"] = data["minifigs"].fillna(0)
data["agerange_min"] = data["agerange_min"].fillna(0)
```

```python
data.isnull().sum()
```

```python
mask = data["US_retailPrice"].isna()

data.loc[mask, "US_retailPrice"] = np.random.randint(
    10, 500, size=mask.sum()
)
```

```python
data.isnull().sum()
```

```python
data.head()
```

```python
LegoByPrice=data.groupby('name')['US_retailPrice'].sum().head(10)
```

```python
LegoByPrice.plot(kind='bar',cmap='coolwarm')
```

```python
ThemegroupByPrice=data.groupby('themeGroup')['US_retailPrice'].sum().head()
```

```python
ThemegroupByPrice.plot(kind='pie',autopct='%1.1f%%',cmap='coolwarm')
```

```python
Numbric=data.select_dtypes(include='number')
```

```python
sns.heatmap(Numbric,cmap='coolwarm_r')
```

```python
ThemeByPrice=data.groupby('theme')['US_retailPrice'].sum().head(10)
```

```python
ThemeByPrice.plot(kind='line')
```

```python
corr = data.describe().corr()
```

```python
corr = data.select_dtypes(include="number").corr()

plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
```
